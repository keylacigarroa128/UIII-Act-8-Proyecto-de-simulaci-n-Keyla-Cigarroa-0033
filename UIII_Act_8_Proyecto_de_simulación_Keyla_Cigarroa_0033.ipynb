### Fotografo, camara 📸

```python
# Asistente Fotográfico con Inteligencia Artificial
# Nivel: Principiante (Solo if, elif, else y while)

# Mensaje de bienvenida al usuario
print("==================================================")
print("📸 ¡Bienvenido al Asistente Fotográfico con IA! 📸")
print("==================================================")
print("Este sistema inteligente te ayudará a configurar tu")
print("cámara según las condiciones de tu entorno.")

# Variable para mantener el menú activo
continuar = "si"

# Bucle principal para que el programa sea interactivo y no se cierre
while continuar == "si":
    print("\n--- MENÚ DE ENTRENAMIENTO DE LA IA ---")
    print("1. Modo Inteligente: Analizar entorno actual")
    print("2. Modo Diagnóstico: Evaluar error en la foto")
    print("3. Salir del Asistente")

    opcion = input("\nSelecciona una opción (1-3): ")

    # ----------------------------------------------------
    # OPCIÓN 1: MODO INTELIGENTE (Toma de decisiones)
    # ----------------------------------------------------
    if opcion == "1":
        print("\n--- 🤖 ANALIZADOR DE ENTORNO IA ---")
        print("¿Dónde vas a tomar la foto?")
        print("- exterior_sol (Día soleado en la calle)")
        print("- exterior_noche (Calle de noche o concierto)")
        print("- interior_luz (Habitación bien iluminada)")
        print("- accion (Deportes o mascotas corriendo)")

        entorno = input("\nEscribe el entorno actual de la lista: ")

        # La IA evalúa las variables usando condicionales
        if entorno == "exterior_sol":
            print("\n[IA RECOMIENDA]: Configuración para Paisaje Soleado")
            print("- ISO: 100 (Para evitar ruido digital)")
            print("- Velocidad de obturación: 1/1000s (Mucha luz disponible)")
            print("- Diafragma: f/8 (Para que todo salga nítido)")

        elif entorno == "exterior_noche":
            print("\n[IA RECOMIENDA]: Configuración Nocturna")
            print("- ISO: 3200 (Necesitamos sensibilidad a la luz)")
            print("- Velocidad de obturación: 1/60s (¡Cuidado con el pulso!)")
            print("- Diafragma: f/2.8 (Abre el lente al máximo)")

        elif entorno == "interior_luz":
            print("\n[IA RECOMIENDA]: Configuración de Interior")
            print("- ISO: 400 (Luz moderada)")
            print("- Velocidad de obturación: 1/125s")
            print("- Diafragma: f/4")

        elif entorno == "accion":
            print("\n[IA RECOMIENDA]: Configuración para Deportes/Acción")
            print("- ISO: 800 (Para compensar la velocidad rápida)")
            print("- Velocidad de obturación: 1/2000s (¡Congela el movimiento!)")
            print("- Diafragma: f/4")

        else:
            print("\n❌ Entorno no reconocido. Inténtalo de nuevo con los nombres de la lista.")

    # ----------------------------------------------------
    # OPCIÓN 2: MODO DIAGNÓSTICO (Solución de problemas)
    # ----------------------------------------------------
    elif opcion == "2":
        print("\n--- 🔧 DIAGNÓSTICO DE ERRORES IA ---")
        print("¿Qué problema tiene tu fotografía?")
        print("- borrosa (El objeto se movió o salieron líneas de movimiento)")
        print("- oscura (No se ve casi nada)")
        print("- blanca (Tiene demasiada luz, está quemada)")

        problema = input("\nEscribe el problema de tu foto: ")

        if problema == "borrosa":
            print("\n[IA SOLUCIÓN]: El tiempo de exposición es muy largo.")
            print("👉 Sube la velocidad de obturación (ej. de 1/30s a 1/250s).")

        elif problema == "oscura":
            print("\n[IA SOLUCIÓN]: Falta luz en el sensor.")
            print("👉 Abre más el diafragma (número f más chico), sube el ISO o baja la velocidad.")

        elif problema == "blanca":
            print("\n[IA SOLUCIÓN]: Entró demasiada luz.")
            print("👉 Baja el ISO a 100, cierra el diafragma (número f más grande) o aumenta la velocidad.")

        else:
            print("\n❌ Problema no registrado en la base de datos de la IA.")

    # ----------------------------------------------------
    # OPCIÓN 3: SALIR
    # ----------------------------------------------------
    elif opcion == "3":
        print("\nApagando el Asistente IA... ¡Buenas fotos! 📸")
        continuar = "no" # Cambiamos la variable para romper el ciclo while

    # ----------------------------------------------------
    # CASO DE ERROR EN EL MENÚ PRINCIPAL
    # ----------------------------------------------------
    else:
        print("\n❌ Opción del menú inválida. Por favor, elige 1, 2 o 3.")

    # Pregunta de cortesía para continuar si no eligió salir directamente
    if opcion != "3":
        print("\n--------------------------------------------------")
        continuar = input("¿Deseas hacer otra consulta? (si/no): ")
        if continuar != "si":
            print("\n¡Gracias por usar la IA Fotográfica! Hasta pronto.")
```
